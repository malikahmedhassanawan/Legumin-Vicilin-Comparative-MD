```python
import os
import numpy as np
import matplotlib.pyplot as plt

graphs_dir = '/content/drive/MyDrive/Vicilin_MD/Vicilin_300K_200ps'
os.makedirs(graphs_dir, exist_ok=True)

def load_xvg(path):
    data = []
    with open(path) as f:
        for line in f:
            if line.startswith(('#', '@')):
                continue
            data.append([float(x) for x in line.split()])
    return np.array(data)

run_dir = '/content/run'

# --- RMSD ---
rmsd = load_xvg(f'{run_dir}/rmsd_300K.xvg')
plt.figure(figsize=(7, 4.5))
plt.plot(rmsd[:,0], rmsd[:,1], color='blue', linewidth=1.2)
plt.title('Vicilin Dimer Backbone RMSD (200 ps)')
plt.xlabel('Time (ns)')
plt.ylabel('RMSD (nm)')
plt.tight_layout()
plt.savefig(f'{graphs_dir}/rmsd_300K.png', dpi=300)
plt.show()

# --- Radius of Gyration ---
rg = load_xvg(f'{run_dir}/gyrate_300K.xvg')
plt.figure(figsize=(7, 4.5))
plt.plot(rg[:,0], rg[:,1], color='blue', linewidth=1.2)
plt.title('Radius of Gyration (200 ps)')
plt.xlabel('Time (ns)')
plt.ylabel('Rg (nm)')
plt.tight_layout()
plt.savefig(f'{graphs_dir}/rg_300K.png', dpi=300)
plt.show()

# --- RMSF (two chains) ---
rmsf_a = load_xvg(f'{run_dir}/rmsf_chainA_300K.xvg')
rmsf_b = load_xvg(f'{run_dir}/rmsf_chainB_300K.xvg')
plt.figure(figsize=(10, 5))
plt.plot(rmsf_a[:,0], rmsf_a[:,1], color='blue', linewidth=1.3, linestyle='-', label='Chain A')
plt.plot(rmsf_b[:,0], rmsf_b[:,1], color='red', linewidth=1.3, linestyle='--', label='Chain B')
plt.title('RMSF of Vicilin at 300K')
plt.xlabel('Residue')
plt.ylabel('RMSF (nm)')
plt.legend()
plt.tight_layout()
plt.savefig(f'{graphs_dir}/rmsf_300K.png', dpi=300)
plt.show()

# --- SASA ---
sasa = load_xvg(f'{run_dir}/sasa_300K.xvg')
plt.figure(figsize=(7, 4.5))
plt.plot(sasa[:,0], sasa[:,1], color='blue', linewidth=1.2)
plt.title('SASA — 300K Production (200 ps)')
plt.xlabel('Time (ps)')
plt.ylabel('SASA (nm²)')
plt.tight_layout()
plt.savefig(f'{graphs_dir}/sasa_300K.png', dpi=300)
plt.show()

# --- Hydrogen Bonds ---
hbond = load_xvg(f'{run_dir}/hbnum_300K.xvg')
plt.figure(figsize=(7, 4.5))
plt.plot(hbond[:,0], hbond[:,1], color='blue', linewidth=1.2)
plt.title('Hydrogen Bonds — 300K Production (200 ps)')
plt.xlabel('Time (ps)')
plt.ylabel('Number of H-bonds')
plt.tight_layout()
plt.savefig(f'{graphs_dir}/hbond_300K.png', dpi=300)
plt.show()

print("All graphs saved to:", graphs_dir)
```

```python
import matplotlib.pyplot as plt
import os, shutil

LEG_200 = "/content/drive/MyDrive/Legumin_MD/Xvg files 300K_200ps"
LEG_500 = "/content/drive/MyDrive/Legumin_MD/Xvg files 300K_500ps"
VIC_200 = "/content/drive/MyDrive/Vicilin_MD/Xvg files 300K_200ps"
VIC_500 = "/content/drive/MyDrive/Vicilin_MD/Xvg files 300K_500ps"

OUT = "/content/run/Comparison_Legumin_vs_Vicilin"
os.makedirs(OUT, exist_ok=True)

def read_xvg(path, ns_to_ps=False):
    x, y = [], []
    with open(path) as f:
        for line in f:
            if line.startswith(('#', '@')):
                continue
            parts = line.split()
            if len(parts) >= 2:
                xv = float(parts[0])
                x.append(xv * 1000 if ns_to_ps else xv)
                y.append(float(parts[1]))
    return x, y

# ---- Ensure Legumin 300K_500ps RMSF chain files exist in the target folder ----
leg500_chA = f"{LEG_500}/rmsf_300K_500ps_chA.xvg"
leg500_chB = f"{LEG_500}/rmsf_300K_500ps_chB.xvg"
if not (os.path.exists(leg500_chA) and os.path.exists(leg500_chB)):
    fallback_dir = "/content/drive/MyDrive/Legumin_MD/300k at 500ps"
    fA = f"{fallback_dir}/rmsf_300K_500ps_chA.xvg"
    fB = f"{fallback_dir}/rmsf_300K_500ps_chB.xvg"
    if os.path.exists(fA) and os.path.exists(fB):
        shutil.copy(fA, leg500_chA)
        shutil.copy(fB, leg500_chB)
        print("Copied existing chain-split RMSF into Xvg files 300K_500ps")
    else:
        # Split from combined file as last resort (assumes symmetric homodimer)
        x, y = read_xvg(f"{LEG_500}/rmsf_300K_500ps.xvg")
        half = len(x) // 2
        with open(leg500_chA, 'w') as f:
            for xi, yi in zip(x[:half], y[:half]):
                f.write(f"{xi:.6f} {yi:.6f}\n")
        with open(leg500_chB, 'w') as f:
            for xi, yi in zip(x[half:], y[half:]):
                f.write(f"{(xi - x[half]):.6f} {yi:.6f}\n")
        print("Generated chain-split RMSF from combined file")

def plot_compare_single(leg_path, vic_path, ylabel, title, outpath,
                         leg_ns_to_ps=False, vic_ns_to_ps=False):
    xL, yL = read_xvg(leg_path, ns_to_ps=leg_ns_to_ps)
    xV, yV = read_xvg(vic_path, ns_to_ps=vic_ns_to_ps)
    plt.figure(figsize=(11, 5))
    plt.plot(xL, yL, color='red',  linewidth=1.3, label='Legumin')
    plt.plot(xV, yV, color='blue', linewidth=1.3, label='Vicilin')
    plt.title(title)
    plt.xlabel('Time (ps)')
    plt.ylabel(ylabel)
    plt.legend()
    plt.tight_layout()
    plt.savefig(outpath, dpi=200)
    plt.close()
    print(f"Saved: {outpath}")

def plot_compare_rmsf(legA, legB, vicA, vicB, title, outpath):
    xLA, yLA = read_xvg(legA)
    xLB, yLB = read_xvg(legB)
    xVA, yVA = read_xvg(vicA)
    xVB, yVB = read_xvg(vicB)
    plt.figure(figsize=(11, 5))
    plt.plot(xVA, yVA, color='lightblue', linewidth=1.5, label='Vicilin (Chain A)')
    plt.plot(xVB, yVB, color='lightblue', linewidth=1.5, linestyle='--', label='Vicilin (Chain B)')
    plt.plot(xLA, yLA, color='red', linewidth=1.5, label='Legumin (Chain A)')
    plt.plot(xLB, yLB, color='red', linewidth=1.5, linestyle='--', label='Legumin (Chain B)')
    plt.title(title)
    plt.xlabel('Residue')
    plt.ylabel('RMSF (nm)')
    plt.legend()
    plt.tight_layout()
    plt.savefig(outpath, dpi=200)
    plt.close()
    print(f"Saved: {outpath}")

# ===================== 300K / 200ps =====================
plot_compare_single(f"{LEG_200}/rmsd_300K.xvg", f"{VIC_200}/rmsd_300K_200ps.xvg",
    "RMSD (nm)", "RMSD Comparison at 300K (200ps)",
    f"{OUT}/rmsd_300K_200ps_compare.png", leg_ns_to_ps=True)

plot_compare_single(f"{LEG_200}/gyrate_300K.xvg", f"{VIC_200}/rg_300K_200ps.xvg",
    "Rg (nm)", "Radius of Gyration Comparison at 300K (200ps)",
    f"{OUT}/rg_300K_200ps_compare.png", leg_ns_to_ps=True)

plot_compare_single(f"{LEG_200}/sasa_300K.xvg", f"{VIC_200}/sasa_300K_200ps.xvg",
    "SASA (nm²)", "SASA Comparison at 300K (200ps)",
    f"{OUT}/sasa_300K_200ps_compare.png")

plot_compare_single(f"{LEG_200}/hbnum_300K.xvg", f"{VIC_200}/hbnum_300K_200ps.xvg",
    "H-bond count", "Hydrogen Bonds Comparison at 300K (200ps)",
    f"{OUT}/hbonds_300K_200ps_compare.png")

plot_compare_rmsf(f"{LEG_200}/rmsf_300K_200ps_chA.xvg", f"{LEG_200}/rmsf_300K_200ps_chB.xvg",
    f"{VIC_200}/rmsf_chainA_300K.xvg", f"{VIC_200}/rmsf_chainB_300K.xvg",
    "RMSF Comparison at 300K (200ps)",
    f"{OUT}/rmsf_300K_200ps_compare.png")

# ===================== 300K / 500ps =====================
plot_compare_single(f"{LEG_500}/rmsd_300K_500ps.xvg", f"{VIC_500}/rmsd_500ps.xvg",
    "RMSD (nm)", "RMSD Comparison at 300K (500ps)",
    f"{OUT}/rmsd_300K_500ps_compare.png")

plot_compare_single(f"{LEG_500}/gyrate_300K_500ps.xvg", f"{VIC_500}/gyrate_500ps.xvg",
    "Rg (nm)", "Radius of Gyration Comparison at 300K (500ps)",
    f"{OUT}/rg_300K_500ps_compare.png")

plot_compare_single(f"{LEG_500}/sasa_300K_500ps.xvg", f"{VIC_500}/sasa_500ps.xvg",
    "SASA (nm²)", "SASA Comparison at 300K (500ps)",
    f"{OUT}/sasa_300K_500ps_compare.png")

plot_compare_single(f"{LEG_500}/hbonds_300K_500ps.xvg", f"{VIC_500}/hbnum_500ps.xvg",
    "H-bond count", "Hydrogen Bonds Comparison at 300K (500ps)",
    f"{OUT}/hbonds_300K_500ps_compare.png")

plot_compare_rmsf(leg500_chA, leg500_chB,
    f"{VIC_500}/rmsf_chainA_500ps.xvg", f"{VIC_500}/rmsf_chainB_500ps.xvg",
    "RMSF Comparison at 300K (500ps)",
    f"{OUT}/rmsf_300K_500ps_compare.png")

# ===================== Copy all to Drive =====================
DRIVE_OUT = "/content/drive/MyDrive/Comparison_Legumin_vs_Vicilin"
os.makedirs(DRIVE_OUT, exist_ok=True)
for fname in os.listdir(OUT):
    shutil.copy(f"{OUT}/{fname}", f"{DRIVE_OUT}/{fname}")
print(f"\nAll comparison graphs copied to: {DRIVE_OUT}")
```

```python
import shutil, os

DRIVE_DIR = "/content/drive/MyDrive/Legumin_MD"
TARGET_DIR = f"{DRIVE_DIR}/Xvg files 300K_200ps"

# 1. Copy trajectory + topology to local scratch
os.makedirs('/content/run', exist_ok=True)
shutil.copy(f'{DRIVE_DIR}/md_200ps.tpr', '/content/run/')
shutil.copy(f'{DRIVE_DIR}/md.xtc', '/content/run/')

%cd /content/run

# 2. Re-center the trajectory
!echo -e "1\n0" | gmx trjconv -s md_200ps.tpr -f md.xtc \
    -o md_300K_200ps_center.xtc -pbc mol -center

# 3. RMSD
!echo -e "4\n4" | gmx rms -s md_200ps.tpr -f md_300K_200ps_center.xtc -o rmsd_300K_200ps.xvg -tu ps

# 4. Rg (radius of gyration)
!echo 1 | gmx gyrate -s md_200ps.tpr -f md_300K_200ps_center.xtc -o rg_300K_200ps.xvg

# 5. SASA
!echo 1 | gmx sasa -s md_200ps.tpr -f md_300K_200ps_center.xtc -o sasa_300K_200ps.xvg

# 6. H-bonds
!echo 1 | gmx hbond -s md_200ps.tpr -f md_300K_200ps_center.xtc -num hbnum_300K_200ps.xvg -r "Protein" -t "Protein"

# 7. Combined RMSF
!echo 1 | gmx rmsf -s md_200ps.tpr -f md_300K_200ps_center.xtc -o rmsf_300K_200ps.xvg -res

# 8. Split combined RMSF into Chain A / Chain B (homodimer, split in half)
def read_xvg(path):
    x, y = [], []
    with open(path) as f:
        for line in f:
            if line.startswith(('#', '@')):
                continue
            parts = line.split()
            if len(parts) >= 2:
                x.append(float(parts[0]))
                y.append(float(parts[1]))
    return x, y

resid, rmsf = read_xvg('rmsf_300K_200ps.xvg')
half = len(resid) // 2
print(f"Total residues: {len(resid)} -> Chain A: 1-{half}, Chain B: 1-{len(resid)-half}")

with open('rmsf_300K_200ps_chA.xvg', 'w') as f:
    for i in range(half):
        f.write(f"{i+1:6.0f}   {rmsf[i]:.4f}\n")

with open('rmsf_300K_200ps_chB.xvg', 'w') as f:
    for i in range(half, len(resid)):
        f.write(f"{i-half+1:6.0f}   {rmsf[i]:.4f}\n")

# 9. Copy everything back to Drive
for fname in ['rmsd_300K_200ps.xvg', 'rg_300K_200ps.xvg', 'sasa_300K_200ps.xvg',
              'hbnum_300K_200ps.xvg', 'rmsf_300K_200ps.xvg',
              'rmsf_300K_200ps_chA.xvg', 'rmsf_300K_200ps_chB.xvg']:
    shutil.copy(fname, f'{TARGET_DIR}/{fname}')
    print(f"Saved: {fname}")
```

```python
import matplotlib.pyplot as plt
import os, shutil

LEG_200 = "/content/drive/MyDrive/Legumin_MD/Xvg files 300K_200ps"
VIC_200 = "/content/drive/MyDrive/Vicilin_MD/Xvg files 300K_200ps"

OUT = "/content/run/Comparison_Legumin_vs_Vicilin"
os.makedirs(OUT, exist_ok=True)

def read_xvg(path, ns_to_ps=False):
    x, y = [], []
    with open(path) as f:
        for line in f:
            if line.startswith(('#', '@')):
                continue
            parts = line.split()
            if len(parts) >= 2:
                xv = float(parts[0])
                x.append(xv * 1000 if ns_to_ps else xv)
                y.append(float(parts[1]))
    return x, y

def plot_compare_single(leg_path, vic_path, ylabel, title, outpath,
                         leg_ns_to_ps=False, vic_ns_to_ps=False):
    xL, yL = read_xvg(leg_path, ns_to_ps=leg_ns_to_ps)
    xV, yV = read_xvg(vic_path, ns_to_ps=vic_ns_to_ps)
    plt.figure(figsize=(11, 5))
    plt.plot(xL, yL, color='red',  linewidth=1.3, label='Legumin')
    plt.plot(xV, yV, color='blue', linewidth=1.3, label='Vicilin')
    plt.title(title)
    plt.xlabel('Time (ps)')
    plt.ylabel(ylabel)
    plt.legend()
    plt.tight_layout()
    plt.savefig(outpath, dpi=200)
    plt.close()
    print(f"Saved: {outpath}")

def plot_compare_rmsf(legA, legB, vicA, vicB, title, outpath):
    xLA, yLA = read_xvg(legA)
    xLB, yLB = read_xvg(legB)
    xVA, yVA = read_xvg(vicA)
    xVB, yVB = read_xvg(vicB)
    plt.figure(figsize=(11, 5))
    plt.plot(xVA, yVA, color='lightblue', linewidth=1.5, label='Vicilin (Chain A)')
    plt.plot(xVB, yVB, color='lightblue', linewidth=1.5, linestyle='--', label='Vicilin (Chain B)')
    plt.plot(xLA, yLA, color='red', linewidth=1.5, label='Legumin (Chain A)')
    plt.plot(xLB, yLB, color='red', linewidth=1.5, linestyle='--', label='Legumin (Chain B)')
    plt.title(title)
    plt.xlabel('Residue')
    plt.ylabel('RMSF (nm)')
    plt.legend()
    plt.tight_layout()
    plt.savefig(outpath, dpi=200)
    plt.close()
    print(f"Saved: {outpath}")

# ===================== 300K / 200ps =====================
plot_compare_single(f"{LEG_200}/rmsd_300K_200ps.xvg", f"{VIC_200}/rmsd_300K_200ps.xvg",
    "RMSD (nm)", "RMSD Comparison at 300K (200ps)",
    f"{OUT}/rmsd_300K_200ps_compare.png")

plot_compare_single(f"{LEG_200}/rg_300K_200ps.xvg", f"{VIC_200}/rg_300K_200ps.xvg",
    "Rg (nm)", "Radius of Gyration Comparison at 300K (200ps)",
    f"{OUT}/rg_300K_200ps_compare.png")

plot_compare_single(f"{LEG_200}/sasa_300K_200ps.xvg", f"{VIC_200}/sasa_300K_200ps.xvg",
    "SASA (nm²)", "SASA Comparison at 300K (200ps)",
    f"{OUT}/sasa_300K_200ps_compare.png")

plot_compare_single(f"{LEG_200}/hbnum_300K_200ps.xvg", f"{VIC_200}/hbnum_300K_200ps.xvg",
    "H-bond count", "Hydrogen Bonds Comparison at 300K (200ps)",
    f"{OUT}/hbonds_300K_200ps_compare.png")

plot_compare_rmsf(f"{LEG_200}/rmsf_300K_200ps_chA.xvg", f"{LEG_200}/rmsf_300K_200ps_chB.xvg",
    f"{VIC_200}/rmsf_chainA_300K_200ps.xvg", f"{VIC_200}/rmsf_chainB_300K_200ps.xvg",
    "RMSF Comparison at 300K (200ps)",
    f"{OUT}/rmsf_300K_200ps_compare.png")

# ===================== Copy all to Drive =====================
DRIVE_OUT = "/content/drive/MyDrive/Comparison_Legumin_vs_Vicilin"
os.makedirs(DRIVE_OUT, exist_ok=True)
for fname in os.listdir(OUT):
    shutil.copy(f"{OUT}/{fname}", f"{DRIVE_OUT}/{fname}")
print(f"\nAll 5 comparison graphs copied to: {DRIVE_OUT}")
```

```python
import matplotlib.pyplot as plt
import os, shutil

LEG_400_200 = "/content/drive/MyDrive/Legumin_MD/Xvg files 400K_200ps"
LEG_400_500 = "/content/drive/MyDrive/Legumin_MD/Xvg files 400K_500ps"
VIC_400_200 = "/content/drive/MyDrive/Vicilin_MD/Xvg files 400K_200ps"
VIC_400_500 = "/content/drive/MyDrive/Vicilin_MD/Xvg files 400K_500ps"

OUT = "/content/run/Comparison_400K"
os.makedirs(OUT, exist_ok=True)

def read_xvg(path):
    x, y = [], []
    with open(path) as f:
        for line in f:
            if line.startswith(('#', '@')):
                continue
            parts = line.split()
            if len(parts) >= 2:
                x.append(float(parts[0]))
                y.append(float(parts[1]))
    return x, y

def plot_compare_single(leg_path, vic_path, ylabel, title, outpath):
    xL, yL = read_xvg(leg_path)
    xV, yV = read_xvg(vic_path)

    plt.figure(figsize=(11, 5))
    plt.plot(xL, yL, color='red', linewidth=1.3, label='Legumin')
    plt.plot(xV, yV, color='blue', linewidth=1.3, label='Vicilin')
    plt.title(title)
    plt.xlabel('Time (ps)')
    plt.ylabel(ylabel)
    plt.legend()
    plt.tight_layout()
    plt.savefig(outpath, dpi=200)
    plt.close()
    print(f"Saved: {outpath}")

def plot_compare_rmsf(legA, legB, vicA, vicB, title, outpath):
    xLA, yLA = read_xvg(legA)
    xLB, yLB = read_xvg(legB)
    xVA, yVA = read_xvg(vicA)
    xVB, yVB = read_xvg(vicB)

    plt.figure(figsize=(11, 5))
    plt.plot(xVA, yVA, color='lightblue', linewidth=1.5,
             label='Vicilin (Chain A)')
    plt.plot(xVB, yVB, color='lightblue', linewidth=1.5,
             linestyle='--', label='Vicilin (Chain B)')
    plt.plot(xLA, yLA, color='red', linewidth=1.5,
             label='Legumin (Chain A)')
    plt.plot(xLB, yLB, color='red', linewidth=1.5,
             linestyle='--', label='Legumin (Chain B)')
    plt.title(title)
    plt.xlabel('Residue')
    plt.ylabel('RMSF (nm)')
    plt.legend()
    plt.tight_layout()
    plt.savefig(outpath, dpi=200)
    plt.close()
    print(f"Saved: {outpath}")


# ============================================================
# 400K / 200ps
# ============================================================

plot_compare_single(
    f"{LEG_400_200}/rmsd_400K_200ps.xvg",
    f"{VIC_400_200}/rmsd_400K_200ps.xvg",
    "RMSD (nm)",
    "RMSD Comparison at 400K (200ps)",
    f"{OUT}/rmsd_400K_200ps_compare.png"
)

plot_compare_single(
    f"{LEG_400_200}/rg_400K_200ps.xvg",
    f"{VIC_400_200}/rg_400K_200ps.xvg",
    "Rg (nm)",
    "Radius of Gyration Comparison at 400K (200ps)",
    f"{OUT}/rg_400K_200ps_compare.png"
)

plot_compare_single(
    f"{LEG_400_200}/sasa_400K_200ps.xvg",
    f"{VIC_400_200}/sasa_400K_200ps.xvg",
    "SASA (nm²)",
    "SASA Comparison at 400K (200ps)",
    f"{OUT}/sasa_400K_200ps_compare.png"
)

plot_compare_single(
    f"{LEG_400_200}/hbnum_400K_200ps.xvg",
    f"{VIC_400_200}/hbnum_400K_200ps.xvg",
    "H-bond count",
    "Hydrogen Bonds Comparison at 400K (200ps)",
    f"{OUT}/hbonds_400K_200ps_compare.png"
)

plot_compare_rmsf(
    f"{LEG_400_200}/rmsf_400K_200ps_chA.xvg",
    f"{LEG_400_200}/rmsf_400K_200ps_chB.xvg",
    f"{VIC_400_200}/rmsf_chainA_400K_200ps.xvg",
    f"{VIC_400_200}/rmsf_chainB_400K_200ps.xvg",
    "RMSF Comparison at 400K (200ps)",
    f"{OUT}/rmsf_400K_200ps_compare.png"
)


# ============================================================
# 400K / 500ps
# ============================================================

plot_compare_single(
    f"{LEG_400_500}/rmsd_400K_500ps.xvg",
    f"{VIC_400_500}/rmsd_400K_500ps.xvg",
    "RMSD (nm)",
    "RMSD Comparison at 400K (500ps)",
    f"{OUT}/rmsd_400K_500ps_compare.png"
)

plot_compare_single(
    f"{LEG_400_500}/rg_400K_500ps.xvg",
    f"{VIC_400_500}/gyrate_400K_500ps.xvg",
    "Rg (nm)",
    "Radius of Gyration Comparison at 400K (500ps)",
    f"{OUT}/rg_400K_500ps_compare.png"
)

plot_compare_single(
    f"{LEG_400_500}/sasa_400K_500ps.xvg",
    f"{VIC_400_500}/sasa_400K_500ps.xvg",
    "SASA (nm²)",
    "SASA Comparison at 400K (500ps)",
    f"{OUT}/sasa_400K_500ps_compare.png"
)

plot_compare_single(
    f"{LEG_400_500}/hbnum_400K_500ps.xvg",
    f"{VIC_400_500}/hbnum_400K_500ps.xvg",
    "H-bond count",
    "Hydrogen Bonds Comparison at 400K (500ps)",
    f"{OUT}/hbonds_400K_500ps_compare.png"
)

plot_compare_rmsf(
    f"{LEG_400_500}/rmsf_400K_500ps_chA.xvg",
    f"{LEG_400_500}/rmsf_400K_500ps_chB.xvg",
    f"{VIC_400_500}/rmsf_chainA_400K_500ps.xvg",
    f"{VIC_400_500}/rmsf_chainB_400K_500ps.xvg",
    "RMSF Comparison at 400K (500ps)",
    f"{OUT}/rmsf_400K_500ps_compare.png"
)


# ============================================================
# Copy all graphs to NEW Drive folder
# ============================================================

DRIVE_OUT = "/content/drive/MyDrive/Comparison_400K"
os.makedirs(DRIVE_OUT, exist_ok=True)

for fname in os.listdir(OUT):
    shutil.copy(
        f"{OUT}/{fname}",
        f"{DRIVE_OUT}/{fname}"
    )

print(f"\nAll comparison graphs saved to:")
print(DRIVE_OUT)
```

```python
import matplotlib.pyplot as plt
import os

OUT = "/content/drive/MyDrive/Comparison_Legumin_vs_Vicilin"
os.makedirs(OUT, exist_ok=True)

# =========================
# 400 K / 200 ps — SASA
# =========================
xL, yL = read_xvg(f"{LEG_400_200}/sasa_400K_200ps.xvg")
xV, yV = read_xvg(f"{VIC_400_200}/sasa_400K_200ps.xvg")

yL_rel = [v / yL[0] for v in yL]
yV_rel = [v / yV[0] for v in yV]

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(xL, yL_rel, linewidth=1.5, label="Legumin")
ax.plot(xV, yV_rel, linewidth=1.5, label="Vicilin")
ax.axhline(1.0, linestyle="--", linewidth=1)

ax.set_title("Relative SASA at 400 K (200 ps)")
ax.set_xlabel("Time (ps)")
ax.set_ylabel("Relative SASA (SASA / SASA₀)")
ax.legend()
ax.grid(True, alpha=0.25)
fig.tight_layout()

fig.savefig(
    f"{OUT}/relative_SASA_Legumin_vs_Vicilin_400K_200ps.png",
    dpi=300, bbox_inches="tight"
)
plt.show()
plt.close()


# =========================
# 400 K / 500 ps — SASA
# =========================
xL, yL = read_xvg(f"{LEG_400_500}/sasa_400K_500ps.xvg")
xV, yV = read_xvg(f"{VIC_400_500}/sasa_400K_500ps.xvg")

yL_rel = [v / yL[0] for v in yL]
yV_rel = [v / yV[0] for v in yV]

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(xL, yL_rel, linewidth=1.5, label="Legumin")
ax.plot(xV, yV_rel, linewidth=1.5, label="Vicilin")
ax.axhline(1.0, linestyle="--", linewidth=1)

ax.set_title("Relative SASA at 400 K (500 ps)")
ax.set_xlabel("Time (ps)")
ax.set_ylabel("Relative SASA (SASA / SASA₀)")
ax.legend()
ax.grid(True, alpha=0.25)
fig.tight_layout()

fig.savefig(
    f"{OUT}/relative_SASA_Legumin_vs_Vicilin_400K_500ps.png",
    dpi=300, bbox_inches="tight"
)
plt.show()
plt.close()


# =========================
# 400 K / 200 ps — Rg
# =========================
xL, yL = read_xvg(f"{LEG_400_200}/rg_400K_200ps.xvg")
xV, yV = read_xvg(f"{VIC_400_200}/rg_400K_200ps.xvg")

yL_rel = [v / yL[0] for v in yL]
yV_rel = [v / yV[0] for v in yV]

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(xL, yL_rel, linewidth=1.5, label="Legumin")
ax.plot(xV, yV_rel, linewidth=1.5, label="Vicilin")
ax.axhline(1.0, linestyle="--", linewidth=1)

ax.set_title("Relative Radius of Gyration at 400 K (200 ps)")
ax.set_xlabel("Time (ps)")
ax.set_ylabel("Relative Rg (Rg / Rg₀)")
ax.legend()
ax.grid(True, alpha=0.25)
fig.tight_layout()

fig.savefig(
    f"{OUT}/relative_Rg_Legumin_vs_Vicilin_400K_200ps.png",
    dpi=300, bbox_inches="tight"
)
plt.show()
plt.close()


# =========================
# 400 K / 500 ps — Rg
# =========================
xL, yL = read_xvg(f"{LEG_400_500}/rg_400K_500ps.xvg")
xV, yV = read_xvg(f"{VIC_400_500}/gyrate_400K_500ps.xvg")

yL_rel = [v / yL[0] for v in yL]
yV_rel = [v / yV[0] for v in yV]

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(xL, yL_rel, linewidth=1.5, label="Legumin")
ax.plot(xV, yV_rel, linewidth=1.5, label="Vicilin")
ax.axhline(1.0, linestyle="--", linewidth=1)

ax.set_title("Relative Radius of Gyration at 400 K (500 ps)")
ax.set_xlabel("Time (ps)")
ax.set_ylabel("Relative Rg (Rg / Rg₀)")
ax.legend()
ax.grid(True, alpha=0.25)
fig.tight_layout()

fig.savefig(
    f"{OUT}/relative_Rg_Legumin_vs_Vicilin_400K_500ps.png",
    dpi=300, bbox_inches="tight"
)
plt.show()
plt.close()

print("DONE — 4 relative 400 K graphs saved to:")
print(OUT)
```
